```python
import csv

# Create sample IPL match data
matches = [
    [1, "CSK", "MI", "CSK"],
    [2, "RCB", "KKR", "KKR"],
    [3, "GT", "RR", "GT"],
    [4, "MI", "DC", "MI"],
    [5, "SRH", "PBKS", "SRH"]
]

# Write data to CSV file
with open("ipl_matches.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Write header
    writer.writerow(["match_id", "team1", "team2", "winner"])

    # Write match records
    writer.writerows(matches)

print("ipl_matches.csv created successfully!")
```

```python
import csv

# Read IPL CSV file
with open("ipl_matches.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print("Match", row["match_id"], "Winner:", row["winner"])
```

```python
import json

# Create sample movie data
movies = [
    {
        "title": "Interstellar",
        "year": 2014,
        "rating": 8.7
    },
    {
        "title": "Inception",
        "year": 2010,
        "rating": 8.8
    },
    {
        "title": "3 Idiots",
        "year": 2009,
        "rating": 8.4
    }
]

# Write movie data to JSON file
with open("movies.json", "w") as file:
    json.dump(movies, file, indent=4)

print("movies.json created successfully!")

import json

# Read movies.json
with open("movies.json", "r") as file:
    movies = json.load(file)

# Print title and rating
for movie in movies:
    print("Title:", movie["title"])
    print("Rating:", movie["rating"])
    print()
```

```python
from pathlib import Path
import json

# Define the file path
file_path = Path("my_fav_apps.json")

# Check whether the file exists
if not file_path.exists():

    # Top 3 mobile apps
    apps = [
        {
            "name": "Instagram",
            "category": "Social Media"
        },
        {
            "name": "Zomato",
            "category": "Food Delivery"
        },
        {
            "name": "Paytm",
            "category": "Finance"
        }
    ]

    # Create the JSON file
    with open(file_path, "w") as file:
        json.dump(apps, file, indent=4)

    print("my_fav_apps.json created successfully!")

else:
    print("my_fav_apps.json already exists.")
```
